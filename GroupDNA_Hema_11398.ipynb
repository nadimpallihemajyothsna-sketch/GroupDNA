```python
file_name = "21415815-hostel_bois.txt"

messages = []
system_count = 0
media_count = 0
deleted_count = 0

with open(file_name, "r", encoding="utf-8") as file:
    lines = file.read().splitlines()

for line in lines:
    if line.strip() == "":
        continue

    parts = line.split(" - ", 1)

    # A chat entry should have a timestamp and text after " - "
    if len(parts) != 2:
        system_count += 1
        continue

    timestamp, details = parts

    # Participant messages have "Sender: message"
    sender_parts = details.split(": ", 1)

    if len(sender_parts) != 2:
        system_count += 1
        continue

    sender, text = sender_parts

    if text == "<Media omitted>":
        media_count += 1
    elif text == "This message was deleted":
        deleted_count += 1

    # Keep these messages so they count in each person's total.
    # Exclude media and deleted text later from word and length analysis.
    messages.append({
        "timestamp": timestamp,
        "sender": sender,
        "text": text
    })

participants = set(message["sender"] for message in messages)

print("Messages:", len(messages))
print("Participants:", len(participants), sorted(participants))
print("System notices:", system_count)
print("Media markers:", media_count)
print("Deleted messages:", deleted_count)
print("First parsed message:", messages[0])
```

```python
from datetime import datetime

participant_counts = {}
chat_dates = set()

for message in messages:
    sender = message["sender"]
    date_text = message["timestamp"].split(", ")[0]

    # Count each participant's messages
    if sender not in participant_counts:
        participant_counts[sender] = 0
    participant_counts[sender] += 1

    # Keep each date once
    chat_dates.add(datetime.strptime(date_text, "%d/%m/%y"))

first_date = min(chat_dates)
last_date = max(chat_dates)
total_days = (last_date - first_date).days + 1
total_messages = len(messages)

print("=" * 45)
print("GROUP OVERVIEW")
print("=" * 45)
print("Group        : Hostel Bois 4ever")
print(f"Period       : {first_date.strftime('%d %B %Y')} to {last_date.strftime('%d %B %Y')}")
print(f"Total days   : {total_days}")
print(f"Total messages: {total_messages:,}")
print(f"Participants : {len(participant_counts)}")
print()
print("MESSAGES PER PERSON")

ranked_people = sorted(
    participant_counts.items(),
    key=lambda item: item[1],
    reverse=True
)

for name, count in ranked_people:
    percentage = count / total_messages * 100
    print(f"{name:<8}: {count:>4,} ({percentage:>5.1f}%)")
```

```python
print("ACTIVITY HEATMAP: SHADED BY MESSAGE COUNT")
print("Person   ", end="")

# Label every third hour to keep the display compact
for hour in range(0, 24, 3):
    print(f"{hour:02d} ", end="")
print()

shades = [".", "░", "▒", "█"]

for row_index, person in enumerate(participants):
    row = heatmap[row_index]
    person_max = max(row)

    print(f"{person:<8} ", end="")

    for hour in range(0, 24, 3):
        count = row[hour]

        if person_max == 0 or count == 0:
            shade = "."
        else:
            shade_level = int(count / person_max * 4)
            shade_level = min(shade_level, 3)
            shade = shades[shade_level]

        print(f"{shade}  ", end="")

    print()
```

```python
stop_words = {
    "i", "is", "the", "a", "and", "or", "to", "of",
    "in", "on", "for", "it", "you", "me", "my", "we"
}


word_counts = {}

for message in messages:
    text = message["text"]

    if text == "<Media omitted>" or text == "This message was deleted":
        continue

    words = text.lower().split()

    for word in words:
        # Remove common punctuation from the start and end
        word = word.strip(".,!?;:'\"()[]{}")

        if word == "" or word in stop_words:
            continue

        if word not in word_counts:
            word_counts[word] = 0
        word_counts[word] += 1

top_words = sorted(
    word_counts.items(),
    key=lambda item: item[1],
    reverse=True
)[:10]

print("THIS GROUP'S FAVOURITE WORDS")
for word, count in top_words:
    bar = "█" * (count // 10)
    print(f"{word:<15} {bar} {count}")
```

```python
from datetime import datetime, timedelta

# Convert each message timestamp to a datetime value
for message in messages:
    message["time"] = datetime.strptime(
        message["timestamp"],
        "%d/%m/%y, %H:%M"
    )

# Average time from another person's message to this person's next reply
reply_gaps = {}

for i in range(1, len(messages)):
    previous_message = messages[i - 1]
    current_message = messages[i]

    if previous_message["sender"] != current_message["sender"]:
        person = current_message["sender"]
        gap_minutes = (
            current_message["time"] - previous_message["time"]
        ).total_seconds() / 60

        if person not in reply_gaps:
            reply_gaps[person] = []
        reply_gaps[person].append(gap_minutes)

print("AVERAGE RESPONSE TIMES")
for person in sorted(reply_gaps):
    average_minutes = sum(reply_gaps[person]) / len(reply_gaps[person])
    print(f"{person:<8}: {average_minutes:.1f} minutes")

# Find each person's longest run of days without messages
all_days = sorted(set(message["time"].date() for message in messages))
active_days = {}

for person in set(message["sender"] for message in messages):
    active_days[person] = set(
        message["time"].date()
        for message in messages
        if message["sender"] == person
    )

print("\nLONGEST SILENT STREAKS")
for person in sorted(active_days):
    longest_streak = 0
    current_streak = 0

    for day in all_days:
        if day not in active_days[person]:
            current_streak += 1
            longest_streak = max(longest_streak, current_streak)
        else:
            current_streak = 0

    print(f"{person:<8}: {longest_streak} days")
```

```python
from datetime import datetime

people = sorted(set(message["sender"] for message in messages))
start_date = min(datetime.strptime(m["timestamp"], "%d/%m/%y, %H:%M").date()
                 for m in messages)
end_date = max(datetime.strptime(m["timestamp"], "%d/%m/%y, %H:%M").date()
               for m in messages)
total_days = (end_date - start_date).days + 1

# Group messages by participant
person_messages = {}
for person in people:
    person_messages[person] = [
        m for m in messages if m["sender"] == person
    ]

# Measure consecutive-message bursts
bursts = {person: [] for person in people}
previous_sender = None
burst_size = 0

for message in messages:
    sender = message["sender"]

    if sender == previous_sender:
        burst_size += 1
    else:
        if previous_sender is not None:
            bursts[previous_sender].append(burst_size)
        previous_sender = sender
        burst_size = 1

if previous_sender is not None:
    bursts[previous_sender].append(burst_size)

caring_words = [
    "okay", "safe", "eat", "sleep", "take care",
    "are you", "please", "reminder", "drink water", "don't forget"
]
laugh_words = {"lol", "lmao", "haha", "rofl", "lmfao"}
special_messages = {"<Media omitted>", "This message was deleted"}

archetypes = {}

for person in people:
    person_msgs = person_messages[person]
    real_msgs = [
        m for m in person_msgs if m["text"] not in special_messages
    ]
    total_person_msgs = len(person_msgs)

    # Messages sent between 11 pm and 4:59 am
    night_count = 0
    for message in person_msgs:
        time = datetime.strptime(message["timestamp"], "%d/%m/%y, %H:%M")
        if time.hour >= 23 or time.hour <= 4:
            night_count += 1
    night_percent = night_count / total_person_msgs * 100

    # Average words per regular message
    word_total = sum(len(m["text"].split()) for m in real_msgs)
    average_words = word_total / len(real_msgs) if real_msgs else 0

    # Uppercase messages or messages with at least two exclamation marks
    drama_count = 0
    for message in real_msgs:
        text = message["text"]
        is_long_enough = len(text) >= 3
        if (is_long_enough and text.isupper()) or text.count("!") >= 2:
            drama_count += 1
    drama_percent = drama_count / len(real_msgs) * 100 if real_msgs else 0

    # Caring keyword count
    caring_count = 0
    for message in real_msgs:
        text = message["text"].lower()
        for phrase in caring_words:
            caring_count += text.count(phrase)

    # Percent of messages containing a laughter word
    funny_count = 0
    for message in real_msgs:
        words = message["text"].lower().split()
        words = [word.strip(".,!?;:'\"()[]{}") for word in words]
        if any(word in laugh_words for word in words):
            funny_count += 1
    funny_percent = funny_count / len(real_msgs) * 100 if real_msgs else 0

    # Percent of messages ending in a question mark
    question_count = sum(m["text"].rstrip().endswith("?") for m in real_msgs)
    question_percent = question_count / len(real_msgs) * 100 if real_msgs else 0

    # Percent of days with no messages
    active_days = set()
    for message in person_msgs:
        time = datetime.strptime(message["timestamp"], "%d/%m/%y, %H:%M")
        active_days.add(time.date())
    silent_percent = (total_days - len(active_days)) / total_days * 100

    average_burst = sum(bursts[person]) / len(bursts[person])

    # Scores are scaled by the example thresholds in the brief.
    # The comedian threshold is set at 20% as a project choice.
    scores = {
        "THE SPAMMER": average_burst / 3,
        "THE GROUP MOM": caring_count / 218,
        "THE NIGHT OWL": night_percent / 60,
        "THE STORYTELLER": average_words / 30,
        "THE DRAMA QUEEN": drama_percent / 30,
        "THE GHOST": silent_percent / 60,
        "THE COMEDIAN": funny_percent / 20,
        "THE QUESTION MASTER": question_percent / 25
    }

    best_archetype = max(scores, key=scores.get)
    archetypes[person] = best_archetype

print("PERSONALITY ARCHETYPES")
for person in people:
    print(f"{person:<8} → {archetypes[person]}")
```

```python
from datetime import datetime, timedelta

people = sorted(set(m["sender"] for m in messages))
counts_by_person = {}
counts_by_day = {}
counts_by_hour = {}
active_days = {person: set() for person in people}
reply_gaps = {person: [] for person in people}
dates = []

for message in messages:
    time = datetime.strptime(message["timestamp"], "%d/%m/%y, %H:%M")
    person = message["sender"]
    dates.append(time.date())

    counts_by_person[person] = counts_by_person.get(person, 0) + 1
    counts_by_day[time.date()] = counts_by_day.get(time.date(), 0) + 1
    counts_by_hour[time.hour] = counts_by_hour.get(time.hour, 0) + 1
    active_days[person].add(time.date())

# Calculate reply gaps between consecutive messages from different people
for i in range(1, len(messages)):
    previous = messages[i - 1]
    current = messages[i]

    if previous["sender"] != current["sender"]:
        previous_time = datetime.strptime(
            previous["timestamp"], "%d/%m/%y, %H:%M"
        )
        current_time = datetime.strptime(
            current["timestamp"], "%d/%m/%y, %H:%M"
        )
        gap = (current_time - previous_time).total_seconds() / 60
        reply_gaps[current["sender"]].append(gap)

first_date = min(dates)
last_date = max(dates)
total_days = (last_date - first_date).days + 1
total_messages = len(messages)

busiest_day = max(counts_by_day, key=counts_by_day.get)
busiest_hour = max(counts_by_hour, key=counts_by_hour.get)

# Find the longest no-message streak for each person
longest_streaks = {}

for person in people:
    longest = 0
    current_streak = 0
    day = first_date

    while day <= last_date:
        if day not in active_days[person]:
            current_streak += 1
            longest = max(longest, current_streak)
        else:
            current_streak = 0
        day += timedelta(days=1)

    longest_streaks[person] = longest

# Print the report
line = "=" * 64
print(line)
print('GROUPDNA REPORT — "Hostel Bois 4ever"')
print(f"{total_days} days | {total_messages:,} messages | {len(people)} members")
print(line)

print(f"Period       : {first_date.strftime('%d %B %Y')} to "
      f"{last_date.strftime('%d %B %Y')}")
print(f"Busiest day  : {busiest_day.strftime('%d %B %Y')} "
      f"({counts_by_day[busiest_day]} messages)")
print(f"Busiest hour : {busiest_hour:02d}:00 - {busiest_hour + 1:02d}:00")
print("\nMESSAGES PER PERSON")
for person, count in sorted(
    counts_by_person.items(), key=lambda item: item[1], reverse=True
):
    bar_length = int(count / max(counts_by_person.values()) * 20)
    print(f"{person:<8} {'█' * bar_length} {count} "
          f"({count / total_messages * 100:.1f}%)")

print("\nACTIVITY HEATMAP (sampled every 3 hours)")
print("Person   " + " ".join(f"{hour:02d}" for hour in range(0, 24, 3)))

for row_index, person in enumerate(people):
    row = heatmap[row_index]
    person_max = max(row)
    blocks = []

    for hour in range(0, 24, 3):
        value = row[hour]
        if value == 0 or person_max == 0:
            blocks.append(".")
        elif value / person_max < 0.5:
            blocks.append("░")
        elif value / person_max < 0.75:
            blocks.append("▒")
        else:
            blocks.append("█")

    print(f"{person:<8} " + "  ".join(blocks))

print("\nTHIS GROUP'S FAVOURITE WORDS")
max_word_count = max(count for word, count in top_words)

for word, count in top_words[:10]:
    bar_length = max(1, int(count / max_word_count * 20))
    print(f"{word:<15} {'█' * bar_length} {count}")

print("\nAVERAGE RESPONSE TIMES")
for person in people:
    if reply_gaps[person]:
        average = sum(reply_gaps[person]) / len(reply_gaps[person])
        print(f"{person:<8}: {average:.1f} minutes")

print("\nLONGEST SILENT STREAKS")
for person in sorted(
    longest_streaks, key=longest_streaks.get, reverse=True
):
    print(f"{person:<8}: {longest_streaks[person]} days")

print("\nPERSONALITY ARCHETYPES")
for person in people:
    print(f"{person:<8} → {archetypes[person]}")

print(line)
print("Generated by GroupDNA | Built with Python + NumPy")
print(line)
```
